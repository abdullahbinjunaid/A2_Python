Filling methods

***opening and writing a new file***

`In Write mode`

```python
f = open("file1.txt","w") #write mode
f.write("hello\n") #writing a line
f.write("I am abj\n")
f.close() #closing a file

#for automatice closing of file
with open("file2.txt") as f:
    f.write("this file is opened\n")
    f.write("and closed automatically\n")


```

`In Read Mode`

```python
f = open("file1.txt","r") #read mode
content = f.read() #reads the entire file in one go
print(content)

content = f.readline() #reads a single line from the file
print(content)

data = f.readlines() #For reading all the line of the files and store in the list
print(data)

f.close()
```

`In Append Mode`

```python
f = open("file1.txt","a") #append mode
f.write("this is new line in append mode in the file\n")
f.close()
```
