```python
global StackVowel 
global StackConsonant
StackVowel = []
StackConsonant = []
global VowelTop
global ConsonantTop

VowelTop = -1
ConsonantTop = -1 

def PushData(letter):
    global VowelTop
    global ConsonantTop
    if letter in ["a","e","i","o","u"]:
        if VowelTop == 100:
            print("stack Vowel is full")
        else:
            StackVowel.append(letter)
            VowelTop += 1
            print(f"{letter} added into the stack")
            print(StackVowel)
    else:
        if ConsonantTop == 100:
            print("StackConsonant full")
        else:
            StackConsonant.append(letter)
            ConsonantTop += 1
            print(f"{letter} added into the stack")
            print(StackConsonant)

def PopVowel():
    global VowelTop
    if VowelTop == -1:
        print("stack is empty")
    else:
        value = StackVowel[VowelTop]
        StackVowel[VowelTop] = None
        VowelTop = VowelTop - 1
        print(f"The value which was popped was {value}")
        return value
    
def PopConsonant():
    global ConsonantTop
    if ConsonantTop == -1:
        print("stack is empty")
    else:
        value = StackConsonant[ConsonantTop]
        StackConsonant[ConsonantTop] = None
        ConsonantTop = ConsonantTop - 1
        print(f"The value which was popped was {value}")
        return value

def ReadData():
    file = open("100letters.txt","r")
    letter = file.readline()
    while letter !="":
        PushData(letter.strip())
        letter = file.readline()

    file.close()

ReadData()

print(StackVowel)
print(StackConsonant)
```
