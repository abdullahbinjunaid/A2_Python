```python
#a
DataStored = []   
NumberItems = 0 

#b
def Initialize():
    global DataStored
    global NumberItems
    try:
        NumberItems = int(input("enter the amount of numbers you would like to enter: "))
        while NumberItems > 20 or NumberItems < 1:
            print("out of range please re-enter")
            NumberItems = int(input("enter the amount of numbers you would like to enter: "))
        for i in range(NumberItems):
            num = int(input("enter a number: ")) 
            DataStored.append(num)
            print(DataStored)
    except ValueError:
        print("invalid Data type entered")
    else:
        print(DataStored)
    finally:
        print("program completed")


#c
def BubbleSort():
    global DataStored
    Top = len(DataStored) - 1
    swap = True
    while Top > 1 and swap == True:
        swap = False
        for i in range(Top):
            if DataStored[i]>DataStored[i+1]:
                temp = DataStored[i]
                DataStored[i] = DataStored[i+1]
                DataStored[i+1] = temp 
                swap = True
        Top = Top - 1
#e
def BinarySearch(find):
    global DataStored
    Lb = 0
    Ub = len(DataStored)
    found = False
    
    while (found == False) and (Lb <= Ub):
        Mv = int((Lb+Ub)/2)
        if (find > DataStored[Mv]):
            Lb = Mv + 1
        elif (find < DataStored[Mv]):
            Ub = Mv - 1
        else:
            found = True
    if found == True:
        print(f"{Mv} is the position of {find}")
        return Mv
    else:
        print(f"{find} not found")
        return -1
        


def Main():
    NumberItems = 0
    Initialize()
    BubbleSort()
    print(DataStored)
    find = int(input("enter a number to search in array: "))
    BinarySearch(find)
Main()

```
