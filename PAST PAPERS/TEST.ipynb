```python
class EventItem:
    def __init__(self,EventName, Type, Difficulty):
        self.__EventName = EventName
        self.__Type = Type
        self.__Difficulty = Difficulty
    def GetName(self):
        return self.__EventName
    def GetDifficulty(self):
        return self.__Difficulty
    def GetEventType(self):
        return self.__Type

class Character:
    def __init__(self,CharacterName, Jump, Swim, Run, Drive):
        self.__CharacterName = CharacterName
        self.__Jump = Jump
        self.__Swim = Swim
        self.__Run = Run
        self.__Drive = Drive

    def GetName(self):
        return self.__CharacterName
    def CalculateScore(self, EventType, Diffi):
        chance = 0
        difference = 0
        match EventType:
            case EventType if EventType.lower() == "jump":
                if self.__Jump >= Diffi:
                    chance = 100
                else:
                    difference = Diffi - self.__Jump
            case EventType if EventType.lower() == "swim":
                if self.__Swim >= Diffi:
                    chance = 100
                else:
                    difference = Diffi - self.__Swim
            case EventType if EventType.lower() == "run":
                if self.__Run >= Diffi:
                    chance = 100
                else:
                    difference = Diffi - self.__Run
            case EventType if EventType.lower() == "swim":
                if self.__Swim >= Diffi:
                    chance = 100
                else:
                    difference = Diffi - self.__Swim
            case EventType if EventType.lower() == "drive":
                if self.__Drive >= Diffi:
                    chance = 100
                else:
                    difference = Diffi - self.__Drive

        match difference:
            case difference if difference == 0:
                return chance
            case difference if difference == 1:
                chance = 80
                return chance
            case difference if difference == 2:
                chance = 60
                return chance
            case difference if difference == 3:
                chance = 40
                return chance
            case difference if difference == 4:
                chance = 20
                return chance
    
def main():
    Group = [None] * 5
    Group[0] = EventItem("Bridge", "jump", 3)
    Group[1] = EventItem("Water Wade", "swim", 4)
    Group[2] = EventItem("100 mile run", "run", 5)
    Group[3] = EventItem("Gridlock", "drive", 2)   
    Group[4] = EventItem("Wall on Wall", "jump", 4)

    Tarz = Character("Tarz", 5, 3, 5, 1)
    Geni = Character("Geni", 2, 2, 3, 4)

    TarzScore = 0
    GeniScore = 0

    for event in Group:
        TarzChance = Tarz.CalculateScore(event.GetEventType(), event.GetDifficulty())
        GeniChance = Geni.CalculateScore(event.GetEventType(), event.GetDifficulty())

        print(f"Event: {event.GetName()}")
        print(f"  Tarz: {TarzChance}%  |  Geni: {GeniChance}%")

        if TarzChance > GeniChance:
            TarzScore += 1
            print(f"  {Tarz.GetName()} wins this event!")
        elif GeniChance > TarzChance:
            GeniScore += 1
            print(f"  {Geni.GetName()} wins this event!")
        else:
            print("  This event is a draw.")

    print(f"Final Score — Tarz: {TarzScore}  Geni: {GeniScore}")
    if TarzScore > GeniScore:
        print(f"{Tarz.GetName()} wins the group with {TarzScore} points!")
    elif GeniScore > TarzScore:
        print(f"{Geni.GetName()} wins the group with {GeniScore} points!")
    else:
        print("The group is a draw.")

main()
```
