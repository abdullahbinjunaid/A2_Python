# LINKED LIST

Two pointers are there, start pointer and empty list pointer, they both indicate at the index position of the linked list in a similar way of array or when dealing with lists, the next node beside the the data in a node points towards the next index of the list (both main and free list)

```python
class Node:
  def __init__(self, data, nextNode):
    self.data = data
    self.nextNode = nextNode

LinkedList = [Node(1,1),Node(5,4),Node(6,7),Node(7,-1),Node(2,2),Node(0,6),Node(0,8),Node(56,3),Node(0,9),Node(0,-1)]
startpointer = 0
emptyList = 5

#counting nodes in the link list, along with printing the nodes address and data value
def OUT(LinkedList,startpointer,emptyList):
  Outputpointer = startpointer
  print(f"Data : {LinkedList[Outputpointer].data} , Next Node : {LinkedList[Outputpointer].nextNode}")
  Ncount = 1
  while (LinkedList[Outputpointer].nextNode != -1):
    Outputpointer = LinkedList[Outputpointer].nextNode
    print(f"Data : {LinkedList[Outputpointer].data} , Next Node : {LinkedList[Outputpointer].nextNode}")
    Ncount += 1
  print(Ncount)
  epmtyListPointer = emptyList
  print(f"Data : {LinkedList[epmtyListPointer].data} , Next Node : {LinkedList[epmtyListPointer].nextNode}") # Corrected this line
  Fcount = 1
  while (LinkedList[epmtyListPointer].nextNode != -1):
    epmtyListPointer = LinkedList[epmtyListPointer].nextNode
    print(f"Data : {LinkedList[epmtyListPointer].data} , Next Node : {LinkedList[epmtyListPointer].nextNode}")
    Fcount += 1
  print(Fcount)

OUT(LinkedList,startpointer,emptyList)


#adding a node at the start of the linked list.
temp = startpointer
startpointer = emptyList
emptyList = LinkedList[emptyList].nextNode
LinkedList[startpointer].nextNode = temp

OUT(LinkedList,startpointer,emptyList)

#adding a node in the middle of the list.
outputpointer = startpointer
Ncount = 1
while(Ncount<2):
  outputpointer = LinkedList[outputpointer].nextNode
  Ncount += 1
temp = LinkedList[outputpointer].nextNode
LinkedList[outputpointer].nextNode = emptyList
emptyList = LinkedList[emptyList].nextNode
outputpointer = LinkedList[outputpointer].nextNode
LinkedList[outputpointer].nextNode = temp

OUT(LinkedList,startpointer,emptyList)

#Deleting the node at the start of the list
temp = emptyList
emptyList = startpointer
startpointer = LinkedList[startpointer].nextNode
LinkedList[emptyList].nextNode = temp

OUT(LinkedList,startpointer,emptyList)
```
