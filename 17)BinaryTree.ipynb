# BINARY TREE

There are 3 type of nodes in tree:
1. Root Node
2. Parent Node
3. Child Node
___

### Traversal :

Process of visiting all nodes of a data structure in a specific order


***Three types of travrsals***:
1. `Preorder Traversal` : Root ---> Left ---> Right
2. `Inorder Traversal` : Left ---> Root ---> Right
3. `Postorder Traversal`: Left ---> Right ---> Root

### BINARY-TREE VIA ARRAY AND RECURSION

```python
Tree = ["A","B","C","D","E",None,"F"]
#Preorder Traversal
def Preorder(Tree,index = 0): #initializing the value zero of index at the start of the function 
    if (index >= len(Tree)) or (Tree[index] is None):
        return
    print(Tree[index], end = " ") #Root
    Preorder(Tree, 2 * index + 1) #Left
    Preorder(Tree, 2 * index + 2) #Right.
    
print(Preorder(Tree)) 

#Inorder Traversal
def Inorder(Tree,index = 0):
    if (index >= len(Tree)) or (Tree[index] is None):
        return
    Inorder(Tree, 2 * index + 1) #Left
    print(Tree[index], end = " ") #Root
    Inorder(Tree, 2 * index + 2) #Right.

print(Inorder(Tree))
    
    
    
#Postorder Traversal
def Postorder(Tree,index = 0):
    if (index >= len(Tree)) or (Tree[index] is None):
        return
    Postorder(Tree, 2 * index + 1) #Left
    Postorder(Tree, 2 * index + 2) #Right.
    print(Tree[index], end = " ") #Root
    
    
(Postorder(Tree))
```

### BINARY-TREE VIA OOP AND RECURSION 

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None


# all the nodes are made as objects
root = Node("A")
root.left = Node("B")
root.right = Node("C")
root.left.left = Node("D")
root.left.right = Node("E")
root.right.right = Node("F")

# 1.Preorder Traversal
def Preorder(node):
    if node is not None: #Node != None
        print(node.data, end = " ") #root
        Preorder(node.left) #left
        Preorder(node.right) #right
print(Preorder(root))

# 2.Inorder Traversal
def Inorder(node):
    if node is not None: #Node != None
        Inorder(node.left) #left
        print(node.data, end = " ") #root
        Inorder(node.right) #right
print(Inorder(root))

# 1.Postorder Traversal
def Postorder(node):
    if node is not None: #Node != None
        Postorder(node.left) #left
        Postorder(node.right) #right
        print(node.data, end = " ") #root
print(Postorder(root))
```
