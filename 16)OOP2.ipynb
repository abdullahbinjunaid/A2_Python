# ENCAPSULATION 

`DEF`: Encapsulation is the process of bundling data and the methods that operates into a single unit(class), and restricting direct acess to that data
___

***PURPOSE***

1. Protects data from invalid or unauthorised changes
2. Maintains control over how data is accessed and modified
3. Hides internal implementation details
___

***IN PYTHON***

1. PUBLIC: Accessible everywhere (e.g: self.balance)
2. PROTECTED: Indicated by a single underscore (e.g: self._balance)
3. PRIVATE: Indicated by double underscore (e.g: self.__balance)
___

***KEY IDEA***

Encapsulation ensure that data is accessed only through controlled methods, protecting the intefrity of objects
___

***SUMMARY***

Encapsulation hides data and allows access only through defined methods, ensuring safe and controlled interaction with objects 


```python
# without encapsulation 
class Account:
    def __init__(self, balance):
        self.balance = balance
```

```python
# With Encapsulation 
# 1.Bundling & 2.Access Restriction 
class Account:
    def __init__(self,balance):
        self.__balance = balance
    def deposit(self,amount):
        self.__balance += amount
        
    def get_balance(self):
        return self.__balance
    
acc = Account(1000)
print(acc.get_balance())
#print(acc.__balance) #this is wrong code line as you can't access balance directly you only access it through the method
acc.deposit(1000000)
print(acc.get_balance())
```

# INHERITANCE 

Inheritance is an object-oriented programming concept in which a class can reuse the attributes
and methods of another class. It helps organise code by placing common features in a general
class and allowing more specific classes to build upon it.

`DEF`: Inheritance is the process by which one class acquires the attributes and methods of another class
___

*In many real-world systems, different types of objects share common characteristics. Instead of
rewriting the same code for each class, inheritance allows these shared features to be written once
and reused. This reduces duplication, improves readability, and makes programs easier to
maintain.*
___

***KEY IDEA***

Inheritance represents an 'is-a' relationship. A Teacher is a Person. An Engineer is a Person.
___

***ADVANTAGES***

1. Reduces code duplication
2. Improves program organisation
3. Makes maintenance easier
4. Allows extension of existing classes



```python
#PARENT CLASS
# If encapsulation is done here in parent class then a method is neccesary, to call it, example person's gender
class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def introduce(self):
        print(f"hello my name is {self.name} and my age is {self.age}")
        
#CHILD CLASS        
class Teacher(person):
    def __init__(self,name,age,subject):
        super().__init__(name,age)
        # The function super() is used to call the constructor of the parent class.This ensures that the inherited attributes are properly initialised. 
        self.subject =  subject
        
    def info(self):
        print(self.name ,self.subject)

t1 = Teacher("Abdullah", 100, "maths")
t1.introduce()
print(t1.subject)
t1.info()

class engineer(person):
    def __init__(self, name, age,department):
        super().__init__(name, age)
        self.department = department
    def work(self):
        print(f"{self.name} is working in {self.department}")

class EngTeacher(engineer):
    def __init__(self, name, age, department,subject):
        super().__init__(name, age, department)
        self.subject = subject
    def info(self):
        print(f"this person teaches {self.subject} and works in department {self.department} ")
        
eng1 = EngTeacher("Abdullah", 18 , "coder", "maths")
eng1.info()
```

```python
#OVERRIDING
#A child class can provide its own version of a method that already exists in the parent class. This is
#called method overriding.

class Person:
    def introduce(self):
        print("I am a person.")
class Teacher(Person):
    def introduce(self):
        print("I am a teacher.")

# in this case if Teacher() introduce is called it will override the Person() introduce method
```

# POLYMORPHISM

Polymorphism allows the same method name to behave differently depending on which object is
using it. It is commonly achieved through method overriding in inheritance.

`DEF`: Polymorphism is the ability of the same method to perform different actions depending on the object
that calls it.
___

***KEY IDEA***

The method call remains the same, but the behaviour changes according to the object.
___

***ADVANTAGES***

1. Reduces the need for conditional statements
2. Makes code more flexible and reusable
3. Allows different objects to be treated uniformly
___

***SUMMARY***

Inheritance defines relationships between classes, while polymorphism defines how shared
behaviours can differ. Together, they form a powerful foundation for writing organised and scalable
programs.






```python
class Person:
    def introduce(self):
        print("I am Abdul Samad")
        
class Teacher(Person):
    def intoduce(self):
        print("I am Abdul Samad, a computer science teacher")

class Engineer(Person):
    def introduce(self):
        print("I am Abdul Samad, an electrical engineer")
        
class CSTeacher(Person):
    def introduce(self):
        print("I am Abdul Samad, and I am teaching OOP in Python")
        
people = [Person(), Teacher(), Engineer(), CSTeacher()]
for obj in people:
    obj.introduce()
```
